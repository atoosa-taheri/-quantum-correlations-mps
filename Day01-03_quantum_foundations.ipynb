# 2. Single-Qubit Quantum State

A qubit is the basic unit of quantum information. Unlike a classical bit, which can be either 0 or 1, a qubit can exist in a superposition of the two basis states.

A general pure qubit state is written as:

"|ψ⟩ = α|0⟩ + β|1⟩"

Here:

- "|ψ⟩" represents the quantum state.
- "|0⟩" and "|1⟩" are the computational basis states.
- "α" and "β" are complex probability amplitudes.

For a normalized quantum state, the amplitudes must satisfy:

"|α|² + |β|² = 1"

The probabilities of measuring the qubit in the states "|0⟩" and "|1⟩" are:

- "P(0) = |α|²"
- "P(1) = |β|²"

In this notebook, NumPy arrays represent quantum states as vectors, while "np.linalg.norm()" calculates their norm.

**Key idea**: The amplitudes describe the quantum state, and their squared magnitudes determine measurement probabilities.

```python
import numpy as np

alpha = 1 / np.sqrt(2)
beta = 1j / np.sqrt(2)

P0 = abs(alpha)**2
P1 = abs(beta)**2

print("alpha =", alpha)
print("beta =", beta)
print("P(0) =", P0)
print("P(1) =", P1)
print("Total probability =", P0 + P1)
```

```python
psi = np.array([alpha, beta])

print("Quantum state:")
print(psi)

print("Shape:", psi.shape)
```

```python
np.linalg.norm(psi)
```

```python
psi2 = np.array([2, 1], dtype=complex)

norm2 = np.linalg.norm(psi2)

print("State =", psi2)
print("Norm =", norm2)
```

```python
psi2_normalized = psi2 / norm2

print("Normalized state =", psi2_normalized)
print("New norm =", np.linalg.norm(psi2_normalized))
```

# 4. Quantum Operators: Identity and Pauli Matrices

Quantum operators transform quantum states. In NumPy, an operator is represented as a matrix, and its action on a state is calculated using matrix multiplication.

"|ψ'⟩ = O|ψ⟩"

Here, "O" is an operator, "|ψ⟩" is the initial state, and "|ψ'⟩" is the resulting state.

**Identity operator (I)**

The identity operator leaves a quantum state unchanged:

"I|ψ⟩ = |ψ⟩"

**Pauli operators**

The Pauli matrices are fundamental single-qubit operators:

**- Pauli-X:** swaps the basis states "|0⟩" and "|1⟩".

**- Pauli-Z:** leaves "|0⟩" unchanged and changes the sign of "|1⟩".

**- Pauli-Y:** transforms the basis states with both a sign change and a complex phase.

These operators are also important in quantum circuits and quantum information theory.

In NumPy, the "@" operator performs matrix multiplication, allowing an operator to act on a state vector.

**Key idea:** Quantum operators are matrices that transform state vectors according to the rules of quantum mechanics.

# 5. Quantum Measurement and Probability Amplitudes

Quantum measurement connects a quantum state to the probabilities of observing different outcomes.

For a normalized state "|ψ⟩" and a measurement basis state "|φ⟩", the probability amplitude is the inner product:

"⟨φ|ψ⟩"

The probability of obtaining that measurement outcome is the squared magnitude of the amplitude:

"P(φ) = |⟨φ|ψ⟩|²"

**Important distinction**

**- Probability amplitude:** can be a complex number.

**- Probability:** is a real number between 0 and 1.

For a complete orthonormal measurement basis, the probabilities of all possible outcomes sum to 1.

A projection operator can be constructed from a normalized state "|φ⟩" as:

"Pφ = |φ⟩⟨φ|"

Applying this operator projects a state onto the direction represented by "|φ⟩".

In NumPy, "np.vdot()" calculates the inner product, "abs(amplitude)**2" gives the probability, and "np.outer()" can construct a projection operator.

**Key idea:** Quantum amplitudes determine measurement probabilities, while projection operators represent measurements associated with specific states.

```python
X = np.array([[0, 1],
              [1, 0]], dtype=complex)
```

# Day 2 — Tensor Products and Two-Qubit States

A multi-qubit quantum system is constructed by combining the state spaces of individual qubits using the tensor product.

For two qubits, the combined state space has dimension "2 × 2 = 4".

The tensor product of two vectors is denoted by:

"|ψ⟩ ⊗ |φ⟩"

For the computational basis states, the two-qubit basis is:

- "|00⟩ = |0⟩ ⊗ |0⟩"
- "|01⟩ = |0⟩ ⊗ |1⟩"
- "|10⟩ = |1⟩ ⊗ |0⟩"
- "|11⟩ = |1⟩ ⊗ |1⟩"

These four basis states span the full two-qubit state space.

In NumPy, "np.kron()" computes the Kronecker product, which is the matrix operation used here to construct tensor-product states.

**Key idea:** The tensor product combines individual quantum systems into a larger state space, allowing us to represent and study multi-qubit systems.

```python
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)

state_00 = np.kron(zero, zero)

print("|00> =", state_00)
print("Shape =", state_00.shape)
```

```python
state_01 = np.kron(zero, one)

print("|01>", state_01)
print("Shape=", state_01.shape)
```

```python
state_10 = np.kron(one, zero)
state_11 = np.kron(one, one)

print("|10>", state_10)
print("Shape=", state_10.shape)

print("|11>", state_11)
print("Shape=", state_11.shape)
```

# 7. Coefficient Matrices: Product States and Bell States

A two-qubit pure state can be written as a linear combination of the computational basis states:

"|ψ⟩ = c₀₀|00⟩ + c₀₁|01⟩ + c₁₀|10⟩ + c₁₁|11⟩"

The four coefficients can be arranged into a 2 × 2 coefficient matrix:

"C = [[c₀₀, c₀₁], [c₁₀, c₁₁]]"

This matrix provides a convenient way to analyze the structure of a two-qubit state.

**Product state**

A product state can be expressed as two individual qubit states combined using a tensor product. For example:

"|++⟩ = |+⟩ ⊗ |+⟩"

Its coefficient matrix has rank 1.

**Bell state**

The Bell state used in this notebook is:

"|Φ⁺⟩ = (|00⟩ + |11⟩)/√2"

Its coefficient matrix is:

"C_bell = [[1/√2, 0], [0, 1/√2]]"

Unlike a product state, this Bell state cannot be expressed as a tensor product of two individual qubit states. It is entangled.

**Key idea:** The coefficient matrix organizes the amplitudes of a multi-qubit state and provides the starting point for using SVD to investigate entanglement.

```python
# Product state coefficients
c00_prod, c01_prod, c10_prod, c11_prod = 1/2, 1/2, 1/2, 1/2
psi_2q = (c00_prod * state_00 + c01_prod * state_01 +
c10_prod * state_10 + c11_prod * state_11)

# Bell state coefficients
c00_bell, c01_bell, c10_bell, c11_bell = 1/np.sqrt(2), 0, 0, 1/np.sqrt(2)
bell = (c00_bell * state_00 + c01_bell * state_01 +
c10_bell * state_10 + c11_bell * state_11)

# Coefficient matrices (now correctly separated)
C_product = np.array([[c00_prod, c01_prod], [c10_prod, c11_prod]])
C_bell = np.array([[c00_bell, c01_bell], [c10_bell, c11_bell]])

print("C_product =")
print(C_product)
print("\nC_bell =")
print(C_bell)
```

# 8. Singular Value Decomposition and Schmidt Coefficients

Singular Value Decomposition (SVD) is a matrix factorization method that expresses a matrix as a product of three matrices:

"C = U Σ V†"

Here:

- "C" is the coefficient matrix.
- "U" and "V" contain orthonormal singular vectors.
- "Σ" is a diagonal matrix containing the singular values.
- "V†" denotes the conjugate transpose of "V".

The singular values are non-negative and describe the contribution of each component in the decomposition.

For a bipartite pure quantum state, the singular values of its coefficient matrix correspond to its Schmidt coefficients.

**Schmidt rank**

The Schmidt rank is the number of nonzero Schmidt coefficients.

**- Product state:** Schmidt rank = 1.

**- Bell state:** Schmidt rank = 2.

A larger Schmidt rank indicates that more Schmidt components are required to represent the state across the chosen bipartition.

In NumPy, "np.linalg.svd()" returns "U", the singular values, and "Vh", where "Vh" is the conjugate transpose of "V".

**Key idea:** SVD reveals the mathematical structure of a bipartite quantum state, while the Schmidt coefficients and Schmidt rank help characterize its entanglement.

```python
U_product, S_product, Vh_product = np.linalg.svd(C_product)

threshold = 1e-10
schmidt_rank_product = np.sum(S_product > threshold)

print("Product singular values =", S_product)
print("Product Schmidt rank =", schmidt_rank_product)
print("U_product =")
print(U_product)
print("Vh_product =")
print(Vh_product)
```

```python
U_bell, S_bell, Vh_bell = np.linalg.svd(C_bell)
print("Singular values =", S_bell )
```

# 9. Density Matrices and Global Purity

A density matrix provides a representation of a quantum state and is especially useful when studying subsystems of a larger quantum system.

For a pure state "|ψ⟩", the density matrix is defined as:

"ρ = |ψ⟩⟨ψ|"

For a normalized state, the density matrix has trace 1:

"Tr(ρ) = 1"

**Purity**

Purity measures how pure a quantum state is. It is calculated as:

"γ = Tr(ρ²)"

For a normalized density matrix:

**- Pure state:** purity = 1.

**- Mixed state:** purity is less than 1.

In this notebook, the density matrix is constructed using "np.outer()", and its purity is calculated using matrix multiplication and "np.trace()".

The Bell state is globally pure, even though each individual qubit has a mixed reduced state when considered separately.

**Key idea:** Global purity describes the full quantum state, while reduced purity can reveal how a subsystem is related to the rest of the system.

```python
rho_bell = np.outer(bell, bell.conj())

print("Density matrix =")
print(rho_bell)

print("Shape =", rho_bell.shape)
```

```python
trace_rho = np.trace(rho_bell)

print("Trace =", trace_rho)
```

```python
purity_bell = np.trace(rho_bell @ rho_bell)

print("Global purity of Bell =", purity_bell)
```

# 10. Partial Trace and Reduced Density Matrices

When a quantum system contains multiple subsystems, we may want to describe only one subsystem rather than the entire system.

The **partial trace** is an operation that removes the mathematical description of the subsystem we are not focusing on, producing a reduced density matrix for the subsystem of interest.

For a two-qubit system with density matrix "ρ_AB", the reduced density matrix of subsystem A is:

"ρ_A = Tr_B(ρ_AB)"

Here, "Tr_B" means taking the trace over subsystem B.

**Why is this useful?**

- It allows us to study one part of a larger quantum system.
- It helps distinguish global properties from subsystem properties.
- It is essential for analyzing entanglement.

For the Bell state "|Φ⁺⟩", the reduced density matrix of either qubit is:

"ρ_A = [[1/2, 0], [0, 1/2]]"

This reduced state has trace 1 and purity 0.5, even though the complete Bell state is pure.

In NumPy, the two-qubit density matrix is reshaped into four indices, and "np.trace()" is used to trace out one subsystem.

**Key idea:** The partial trace connects the description of a complete quantum system to the properties of its individual subsystems.

```python
rho_tensor = rho_bell.reshape(2, 2, 2, 2)

print("Shape =", rho_tensor.shape)
```

```python
rho_A = np.trace(rho_tensor, axis1=1, axis2=3)

print("Reduced density matrix:")
print(rho_A)

print("Shape =", rho_A.shape)
```

# 11. Schmidt Rank and Entanglement Structure

The Schmidt rank describes how many nonzero Schmidt coefficients are needed to represent a bipartite pure quantum state.

For a two-qubit pure state, the Schmidt coefficients are obtained from the singular values of its coefficient matrix.

**Interpretation**

**- Schmidt rank = 1:** The state is a product state and contains no entanglement between the two subsystems.

**- Schmidt rank greater than 1:** The state is entangled across the chosen bipartition.

For the states studied in this notebook:

- The product state "|++⟩" has Schmidt rank 1.
- The Bell state "|Φ⁺⟩" has Schmidt rank 2 and is maximally entangled.

In numerical calculations, very small singular values may arise from floating-point errors. A threshold can be used to distinguish meaningful nonzero singular values from values that are effectively zero.

**Key idea:** Schmidt rank describes the number of independent Schmidt components needed to represent a bipartite pure state, making it a useful indicator of entanglement structure.

# 12. Entanglement Entropy: Product vs. Bell States

Entanglement entropy quantifies the entanglement between two subsystems of a pure quantum state.

For subsystem A, the von Neumann entropy is calculated from its reduced density matrix:

"S(ρ_A) = -Tr(ρ_A log₂ ρ_A)"

Equivalently, using the eigenvalues "p_i" of the reduced density matrix:

"S(ρ_A) = -Σ p_i log₂(p_i)"

Here, each "p_i" is a non-negative eigenvalue, and the eigenvalues sum to 1.

**Interpretation**

**- Product state:** The reduced state is pure, so the entanglement entropy is 0 bits.

**- Bell state:** Each individual qubit has a maximally mixed reduced state, giving an entanglement entropy of 1 bit.

The entropy is calculated in bits because the logarithm uses base 2. Eigenvalues equal to zero contribute zero to the entropy and should be excluded from the numerical logarithm calculation.

**Key idea:** Entanglement entropy measures how much entanglement exists between subsystems, distinguishing product states from entangled states.

```python
rho_product_tensor = rho_product.reshape(2, 2, 2, 2)

rho_A_product = np.trace(rho_product_tensor,
axis1=1,axis2=3)

print("rho_A Product =")
print(rho_A_product)
purity_A_product = np.trace(
rho_A_product @ rho_A_product)
print("Reduced purity of Product =", purity_A_product)
```

```python
purity_A_bell = np.trace(rho_A_bell @ rho_A_bell)

print("Reduced purity of Bell =", purity_A_bell)
```
