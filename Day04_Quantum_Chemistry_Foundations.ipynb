
# Day 4 — Molecular Quantum Chemistry with PySCF

**PySCF** is used to calculate the electronic energy of the hydrogen molecule (H₂) using the Restricted Hartree–Fock (RHF) method.

**Molecular Hamiltonian:** Under the Born–Oppenheimer approximation, nuclei are treated as fixed while the electronic Schrödinger equation is solved.

**Hartree–Fock (HF):** Approximates the many-electron wavefunction using a single Slater determinant and treats electron interactions through an averaged mean-field potential.

**Basis Set:** STO-3G is a minimal basis set that is computationally inexpensive but limited in accuracy.

**Electron Correlation:** Hartree–Fock does not fully account for electron correlation. This limitation becomes particularly important as a chemical bond is stretched.

**Goal:** Compare H₂ at two bond distances, 1.4 and 2.0 Bohr, as a starting point for studying electron correlation and tensor network methods.

```python
import pyscf
print("PySCF version:", pyscf.__version__)
```

```python
from pyscf import gto

mol = gto.M(
atom = '''
H 0 0 0
H 0 0 1.4
''',
basis = 'sto-3g',
unit = 'Bohr',
charge = 0,
spin = 0)

print("Number of electrons:", mol.nelectron)
print("Number of basis functions:", mol.nao_nr())
```

```python
from pyscf import scf

mf = scf.RHF(mol)
energy = mf.kernel()

print("HF energy =", energy)
```

```python
mol_stretched = gto.M(
atom = '''
H 0 0 0
H 0 0 2.0
''',
basis = 'sto-3g',
unit = 'Bohr',
charge = 0,
spin = 0)
print("Number of electrons:", mol_stretched.nelectron)
print("Number of basis functions:", mol_stretched.nao_nr())
```

```python
mf_stretched = scf.RHF(mol_stretched)
energy_stretched = mf_stretched.kernel()

print("HF energy at R = 2.0 Bohr =", energy_stretched)
```
